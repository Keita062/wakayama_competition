```python
import os
from pathlib import Path
import pandas as pd

# 相対パスの指定（1つ上の層の data フォルダを参照）
file_path = Path("../data/地域幸福度_市区町村別データ_2022-2026.xlsx")

if not file_path.exists():
    print(
        f"❌ ファイルが見つかりません。絶対パス:\n   {file_path.resolve()}"
    )
    print(f"   現在の実行位置: {os.getcwd()}")
else:
    print("✅ ファイルを確認しました。")

    excel_file = pd.ExcelFile(file_path)
    sheet_names = excel_file.sheet_names
    print("シート一覧:", sheet_names)

    # 2個目のシート（インデックス: 1）
    if len(sheet_names) > 1:
        sheet2_name = sheet_names[1]
        df_sheet2 = pd.read_excel(file_path, sheet_name=1)
        print(f"\n=== 📄 2個目のシート: 【{sheet2_name}】 ===")
        print(f"・サイズ (行, 列): {df_sheet2.shape}")
        display(df_sheet2.head())
    else:
        print("\n⚠️ 2個目のシートが存在しません。")

    # 3個目のシート（インデックス: 2）
    if len(sheet_names) > 2:
        sheet3_name = sheet_names[2]
        df_sheet3 = pd.read_excel(file_path, sheet_name=2)
        print(f"\n=== 📄 3個目のシート: 【{sheet3_name}】 ===")
        print(f"・サイズ (行, 列): {df_sheet3.shape}")
        display(df_sheet3.head())
    else:
        print("\n⚠️ 3個目のシートが存在しません。")
```

```python
# 区分ごとにグループ化して欠損状況を集計
for kbn, group in df_sheet2.groupby("区分"):
    print(f"\n==========================================")
    print(f"■ 区分: 【{kbn}】 (総行数: {len(group)} 行)")
    print(f"==========================================")

    # 欠損がある列のみ抽出
    null_counts = group.isnull().sum()
    null_rates = (group.isnull().mean() * 100).round(2)

    df_kbn_nan = pd.DataFrame({"欠損数": null_counts, "欠損率(%)": null_rates})
    df_kbn_nan_filtered = df_kbn_nan[df_kbn_nan["欠損数"] > 0]

    if df_kbn_nan_filtered.empty:
        print("  ✅ 欠損値（NaN）はありません。")
    else:
        display(df_kbn_nan_filtered)
```

```python
#％が違う原因を調べる、年度によってない質問があるのか？
# 区分「全体」のデータのみ抽出
df_all = df_sheet2[df_sheet2["区分"] == "全体"].copy()

# 欠損数が異なる代表的な2つの列を設定
col_a = "幸福度"  # 欠損数 4,241件（欠損率 47.22%）
col_b = "主観_デジタル生活"  # 欠損数 5,574件（欠損率 62.06%）

# 1. 年度ごとに「数値が入っている（非欠損の）行数」を比較
check_by_year = df_all.groupby("年度")[[col_a, col_b]].count()
check_by_year.columns = [f"{col_a}（データ数）", f"{col_b}（データ数）"]
check_by_year["差分（件数）"] = (
    check_by_year[f"{col_a}（データ数）"]
    - check_by_year[f"{col_b}（データ数）"]
)

print("=== 📄 1. 年度ごとの指標別・データ存在件数 ===")
display(check_by_year)

# 2. 「幸福度」はあるのに「主観_デジタル生活」が欠損している1,333件の年度内訳
diff_rows = df_all[df_all[col_a].notnull() & df_all[col_b].isnull()]
print(
    f"\n=== 📄 2. 『{col_a}』はあり『{col_b}』が無い行（計 {len(diff_rows)} 件）の年度別内訳 ==="
)
print(diff_rows["年度"].value_counts().sort_index())
```

```python
cols = [
    "主観_遊び・娯楽",
    "主観_子育て",
    "主観_初等・中等教育",
    "主観_デジタル生活",
    "主観_自然の恵み",
    "主観_環境共生",
    "主観_自然災害",
    "主観_教育機会の豊かさ",
    "主観_雇用・所得",
    "5年後の幸福度"
]

# 10項目すべてが欠損している行数
all_missing = df_sheet2[cols].isna().all(axis=1).sum()

# 10項目のどれか1つでも欠損している行数
any_missing = df_sheet2[cols].isna().any(axis=1).sum()

print("10項目すべて欠損:", all_missing)
print("10項目のいずれか欠損:", any_missing)
```

```python
# 区分 × 年度の行数
year_counts = (
    df_sheet2
    .groupby(["区分", "年度"])
    .size()
    .reset_index(name="行数")
)

# 区分ごとの総行数
total_by_kbn = year_counts.groupby("区分")["行数"].transform("sum")

# 区分内構成比
year_counts["区分内構成比(%)"] = (
    year_counts["行数"] / total_by_kbn * 100
).round(2)

# 並び順を指定
kbn_order = ["全体", "年代別", "性別別", "年代×性別"]
year_counts["区分"] = pd.Categorical(
    year_counts["区分"],
    categories=kbn_order,
    ordered=True
)

year_counts = (
    year_counts
    .sort_values(["区分", "年度"])
    .reset_index(drop=True)
)

display(year_counts)
```

```python
# 分析対象の項目
target_cols = [
    c for c in df_sheet2.columns
    if c.startswith("主観_")
    or c in ["幸福度", "生活満足度", "5年後の幸福度"]
]

# 区分 × 年度ごとの欠損数・欠損率
missing_counts = []

for (kbn, year), group in df_sheet2.groupby(["区分", "年度"]):
    # 対象項目の欠損セル数
    missing_count = group[target_cols].isna().sum().sum()

    # 対象項目の全セル数
    total_count = group[target_cols].size

    # 欠損率
    missing_rate = missing_count / total_count * 100

    missing_counts.append({
        "区分": kbn,
        "年度": year,
        "欠損数": missing_count,
        "欠損率(%)": round(missing_rate, 2)
    })

missing_by_kbn_year = pd.DataFrame(missing_counts)

# 並び順
kbn_order = ["全体", "年代別", "性別別", "年代×性別"]

missing_by_kbn_year["区分"] = pd.Categorical(
    missing_by_kbn_year["区分"],
    categories=kbn_order,
    ordered=True
)

missing_by_kbn_year = (
    missing_by_kbn_year
    .sort_values(["区分", "年度"])
    .reset_index(drop=True)
)

display(missing_by_kbn_year)
```

```python
import pandas as pd

# 主観データのみ
df = df_sheet2.copy()

# 回答者数を数値化
df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)

# 2022年度を除外
df = df[df["年度"] != 2022].copy()

# 区分ごとの基本統計
summary = (
    df.groupby("区分")["回答者数"]
    .describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])
    .round(1)
)

display(summary)
```

```python
import pandas as pd

df = df_sheet2.copy()

# キー：自治体 × 年度
keys = ["年度", "都道府県", "市区町村", "地域ID"]

# -----------------------------
# 1. 全体データ
# -----------------------------
df_overall = (
    df[df["区分"] == "全体"][keys]
    .drop_duplicates()
)

# -----------------------------
# 2. 性別別データ
# -----------------------------
df_gender = (
    df[df["区分"] == "性別別"][keys + ["性別"]]
    .dropna(subset=["性別"])
    .drop_duplicates()
)

print("性別の値：")
print(df_gender["性別"].value_counts(dropna=False))

# 自治体×年度ごとの性別行数
gender_count = (
    df_gender
    .groupby(keys)["性別"]
    .nunique()
    .reset_index(name="性別行数")
)

# 全体データと結合
check = df_overall.merge(
    gender_count,
    on=keys,
    how="left"
)

check["性別行数"] = check["性別行数"].fillna(0).astype(int)

# -----------------------------
# 3. 性別データの存在状況
# -----------------------------
print("\n=== 性別行数の分布 ===")
print(
    check["性別行数"]
    .value_counts()
    .sort_index()
)

# -----------------------------
# 4. 全体はあるが性別データが不足しているケース
# -----------------------------
missing_gender = check[
    check["性別行数"] < 2
].copy()

print("\n=== 全体はあるが、男女2行そろっていない自治体×年度 ===")
print(f"件数：{len(missing_gender)}")

display(
    missing_gender.sort_values(
        ["年度", "都道府県", "市区町村"]
    ).head(50)
)
```

```python
thresholds = [5, 10, 20, 30, 50]

rows = []

for kubun, group in df.groupby("区分"):
    n_total = group["回答者数"].notna().sum()

    row = {
        "区分": kubun,
        "総行数": n_total
    }

    for threshold in thresholds:
        row[f"n<{threshold}の割合(%)"] = (
            (group["回答者数"] < threshold).mean() * 100
        )

    rows.append(row)

low_n_summary = pd.DataFrame(rows)

display(low_n_summary.round(1))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from pathlib import Path

# =========================
# 1. 日本語フォントを設定
# =========================
jp_fonts = [
    "Noto Sans CJK JP",
    "Noto Sans JP",
    "IPAexGothic",
    "IPAGothic",
    "Yu Gothic",
    "Meiryo"
]

available_fonts = {f.name for f in fm.fontManager.ttflist}

for font_name in jp_fonts:
    if font_name in available_fonts:
        plt.rcParams["font.family"] = font_name
        break

# マイナス記号の文字化け対策
plt.rcParams["axes.unicode_minus"] = False


# =========================
# 2. 分析対象を作成
# =========================
df = df_sheet2.copy()

df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)

df["幸福度"] = pd.to_numeric(
    df["幸福度"],
    errors="coerce"
)

# 2023～2026年
# 区分 = 全体
# 幸福度あり
# 回答者数 > 0
df_analysis = df[
    (df["年度"].between(2023, 2026)) &
    (df["区分"] == "全体") &
    (df["幸福度"].notna()) &
    (df["回答者数"] > 0)
].copy()


# =========================
# 3. 回答者数を取得
# =========================
n = df_analysis["回答者数"].dropna()


# =========================
# 4. グラフ作成
# =========================
fig, ax = plt.subplots(figsize=(10, 6))

# 回答者数の分布
bins = np.logspace(
    np.log10(n.min()),
    np.log10(n.max()),
    30
)

ax.hist(
    n,
    bins=bins
)

# 基準候補を表示
thresholds = [10, 30, 50, 80, 100]
threshold_colors = ["red", "orange", "green", "blue", "purple"]

for threshold, color in zip(thresholds, threshold_colors):
    ax.axvline(
        threshold,
        linestyle="--",
        color=color,
        label=f"n = {threshold}"
    )

# 対数軸
ax.set_xscale("log")

ax.set_xlabel("回答者数")
ax.set_ylabel("観測数")

ax.set_title(
    f"回答者数の分布（2023～2026年・全体区分、n = {len(n):,}）"
)

ax.legend()
ax.grid(
    True,
    which="both",
    axis="both",
    alpha=0.3
)

plt.tight_layout()
# =========================
# 5. nの基準値ごとの件数・割合
# =========================

thresholds = [10, 30, 50, 80, 100]

results = []

total = len(df_analysis)

for threshold in thresholds:
    count_ge = (df_analysis["回答者数"] >= threshold).sum()
    count_lt = (df_analysis["回答者数"] < threshold).sum()

    results.append({
        "基準値": threshold,
        "n以上の観測数": count_ge,
        "n以上の割合(%)": count_ge / total * 100,
        "n未満の観測数": count_lt,
        "n未満の割合(%)": count_lt / total * 100
    })

threshold_summary = pd.DataFrame(results)

threshold_summary["n以上の割合(%)"] = threshold_summary["n以上の割合(%)"].round(1)
threshold_summary["n未満の割合(%)"] = threshold_summary["n未満の割合(%)"].round(1)

display(threshold_summary)
```
