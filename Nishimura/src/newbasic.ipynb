```python
import os
from pathlib import Path
import pandas as pd

# 相対パスの指定（1つ上の層の data フォルダを参照）
file_path = Path("../data/地域幸福度_市区町村別データ_2022-2026.xlsx")

if not file_path.exists():
    print(
        f"❌ ファイルが見つかりません。絶対パス:\n   {file_path.resolve()}"
    )
    print(f"   現在の実行位置: {os.getcwd()}")
else:
    print("✅ ファイルを確認しました。")

    excel_file = pd.ExcelFile(file_path)
    sheet_names = excel_file.sheet_names
    print("シート一覧:", sheet_names)

    # 2個目のシート（インデックス: 1）
    if len(sheet_names) > 1:
        sheet2_name = sheet_names[1]
        df_sheet2 = pd.read_excel(file_path, sheet_name=1)
        print(f"\n=== 📄 2個目のシート: 【{sheet2_name}】 ===")
        print(f"・サイズ (行, 列): {df_sheet2.shape}")
        display(df_sheet2.head())
    else:
        print("\n⚠️ 2個目のシートが存在しません。")

    # 3個目のシート（インデックス: 2）
    if len(sheet_names) > 2:
        sheet3_name = sheet_names[2]
        df_sheet3 = pd.read_excel(file_path, sheet_name=2)
        print(f"\n=== 📄 3個目のシート: 【{sheet3_name}】 ===")
        print(f"・サイズ (行, 列): {df_sheet3.shape}")
        display(df_sheet3.head())
    else:
        print("\n⚠️ 3個目のシートが存在しません。")
```

```python
# 2個目のシート（主観データ）の全列名
print(
    f"=== 📄 2個目のシート【主観データ】の列一覧 (全 {len(df_sheet2.columns)} 列) ==="
)
print(list(df_sheet2.columns))

print("\n" + "=" * 60 + "\n")

# 3個目のシート（客観データ）の全列名
print(
    f"=== 📄 3個目のシート【客観データ】の列一覧 (全 {len(df_sheet3.columns)} 列) ==="
)
print(list(df_sheet3.columns))
```

```python
# 主観データの「区分」内訳を確認
print("【主観データの区分一覧】")
print(df_sheet2["区分"].value_counts())
```

```python
# 区分ごとにグループ化して欠損状況を集計
for kbn, group in df_sheet2.groupby("区分"):
    print(f"\n==========================================")
    print(f"■ 区分: 【{kbn}】 (総行数: {len(group)} 行)")
    print(f"==========================================")

    # 欠損がある列のみ抽出
    null_counts = group.isnull().sum()
    null_rates = (group.isnull().mean() * 100).round(2)

    df_kbn_nan = pd.DataFrame({"欠損数": null_counts, "欠損率(%)": null_rates})
    df_kbn_nan_filtered = df_kbn_nan[df_kbn_nan["欠損数"] > 0]

    if df_kbn_nan_filtered.empty:
        print("  ✅ 欠損値（NaN）はありません。")
    else:
        display(df_kbn_nan_filtered)
```

```python
#％が違う原因を調べる、年度によってない質問があるのか？
# 区分「全体」のデータのみ抽出
df_all = df_sheet2[df_sheet2["区分"] == "全体"].copy()

# 欠損数が異なる代表的な2つの列を設定
col_a = "幸福度"  # 欠損数 4,241件（欠損率 47.22%）
col_b = "主観_デジタル生活"  # 欠損数 5,574件（欠損率 62.06%）

# 1. 年度ごとに「数値が入っている（非欠損の）行数」を比較
check_by_year = df_all.groupby("年度")[[col_a, col_b]].count()
check_by_year.columns = [f"{col_a}（データ数）", f"{col_b}（データ数）"]
check_by_year["差分（件数）"] = (
    check_by_year[f"{col_a}（データ数）"]
    - check_by_year[f"{col_b}（データ数）"]
)

print("=== 📄 1. 年度ごとの指標別・データ存在件数 ===")
display(check_by_year)

# 2. 「幸福度」はあるのに「主観_デジタル生活」が欠損している1,333件の年度内訳
diff_rows = df_all[df_all[col_a].notnull() & df_all[col_b].isnull()]
print(
    f"\n=== 📄 2. 『{col_a}』はあり『{col_b}』が無い行（計 {len(diff_rows)} 件）の年度別内訳 ==="
)
print(diff_rows["年度"].value_counts().sort_index())
```

```python
# 「全体」データのみ
df_all = df_sheet2[df_sheet2["区分"] == "全体"].copy()

# 分析対象となる指標列
indicator_cols = [
    c for c in df_all.columns
    if c.startswith("主観_") or c in ["幸福度", "生活満足度", "5年後の幸福度"]
]

# 年度ごとの欠損率
missing_by_year = (
    df_all.groupby("年度")[indicator_cols]
    .apply(lambda x: x.isna().mean() * 100)
    .round(2)
)

display(missing_by_year)
```

```python
for kbn in df_sheet2["区分"].unique():
    print(f"\n{'=' * 60}")
    print(f"■ 区分：{kbn}")
    print(f"{'=' * 60}")

    display(
        missing_by_kbn_year[
            missing_by_kbn_year["区分"] == kbn
        ][["年度", "項目", "欠損数", "欠損率(%)"]]
        .sort_values(["年度", "項目"])
        .reset_index(drop=True)
    )
```

```python
cols = [
    "主観_遊び・娯楽",
    "主観_子育て",
    "主観_初等・中等教育",
    "主観_デジタル生活",
    "主観_自然の恵み",
    "主観_環境共生",
    "主観_自然災害",
    "主観_教育機会の豊かさ",
    "主観_雇用・所得",
    "5年後の幸福度"
]

# 10項目すべてが欠損している行数
all_missing = df_sheet2[cols].isna().all(axis=1).sum()

# 10項目のどれか1つでも欠損している行数
any_missing = df_sheet2[cols].isna().any(axis=1).sum()

print("10項目すべて欠損:", all_missing)
print("10項目のいずれか欠損:", any_missing)
```

```python
# 区分 × 年度の行数
year_counts = (
    df_sheet2
    .groupby(["区分", "年度"])
    .size()
    .reset_index(name="行数")
)

# 区分ごとの総行数
total_by_kbn = year_counts.groupby("区分")["行数"].transform("sum")

# 区分内構成比
year_counts["区分内構成比(%)"] = (
    year_counts["行数"] / total_by_kbn * 100
).round(2)

# 並び順を指定
kbn_order = ["全体", "年代別", "性別別", "年代×性別"]
year_counts["区分"] = pd.Categorical(
    year_counts["区分"],
    categories=kbn_order,
    ordered=True
)

year_counts = (
    year_counts
    .sort_values(["区分", "年度"])
    .reset_index(drop=True)
)

display(year_counts)
```

```python
# 分析対象の項目
target_cols = [
    c for c in df_sheet2.columns
    if c.startswith("主観_")
    or c in ["幸福度", "生活満足度", "5年後の幸福度"]
]

# 区分 × 年度ごとの欠損数・欠損率
missing_counts = []

for (kbn, year), group in df_sheet2.groupby(["区分", "年度"]):
    # 対象項目の欠損セル数
    missing_count = group[target_cols].isna().sum().sum()

    # 対象項目の全セル数
    total_count = group[target_cols].size

    # 欠損率
    missing_rate = missing_count / total_count * 100

    missing_counts.append({
        "区分": kbn,
        "年度": year,
        "欠損数": missing_count,
        "欠損率(%)": round(missing_rate, 2)
    })

missing_by_kbn_year = pd.DataFrame(missing_counts)

# 並び順
kbn_order = ["全体", "年代別", "性別別", "年代×性別"]

missing_by_kbn_year["区分"] = pd.Categorical(
    missing_by_kbn_year["区分"],
    categories=kbn_order,
    ordered=True
)

missing_by_kbn_year = (
    missing_by_kbn_year
    .sort_values(["区分", "年度"])
    .reset_index(drop=True)
)

display(missing_by_kbn_year)
```

```python
# 全体データ
df_all = df_sheet2[df_sheet2["区分"] == "全体"].copy()

# 年代別データ
df_age = df_sheet2[df_sheet2["区分"] == "年代別"].copy()

# 市×年度の一覧
all_city_year = (
    df_all[["年度", "都道府県", "市区町村"]]
    .drop_duplicates()
)

age_city_year = (
    df_age[["年度", "都道府県", "市区町村"]]
    .drop_duplicates()
)

# 全体にはあるが、年代別には存在しない市×年度
missing_age = all_city_year.merge(
    age_city_year,
    on=["年度", "都道府県", "市区町村"],
    how="left",
    indicator=True
)

missing_age = missing_age[
    missing_age["_merge"] == "left_only"
]

display(
    missing_age.groupby("年度")
    .size()
    .reset_index(name="年代別データが存在しない市数")
)
```
