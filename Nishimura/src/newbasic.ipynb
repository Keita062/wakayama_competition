```python
# 分析対象の項目
target_cols = [
    c for c in df_sheet2.columns
    if c.startswith("主観_")
    or c in ["幸福度", "生活満足度", "5年後の幸福度"]
]

# 区分 × 年度ごとの欠損数・欠損率
missing_counts = []

for (kbn, year), group in df_sheet2.groupby(["区分", "年度"]):
    # 対象項目の欠損セル数
    missing_count = group[target_cols].isna().sum().sum()

    # 対象項目の全セル数
    total_count = group[target_cols].size

    # 欠損率
    missing_rate = missing_count / total_count * 100

    missing_counts.append({
        "区分": kbn,
        "年度": year,
        "欠損数": missing_count,
        "欠損率(%)": round(missing_rate, 2)
    })

missing_by_kbn_year = pd.DataFrame(missing_counts)

# 並び順
kbn_order = ["全体", "年代別", "性別別", "年代×性別"]

missing_by_kbn_year["区分"] = pd.Categorical(
    missing_by_kbn_year["区分"],
    categories=kbn_order,
    ordered=True
)

missing_by_kbn_year = (
    missing_by_kbn_year
    .sort_values(["区分", "年度"])
    .reset_index(drop=True)
)

display(missing_by_kbn_year)
```

```python
import pandas as pd

# =========================
# 1. データ準備
# =========================
df = df_sheet2.copy()

# 2022年度を除外
df = df[df["年度"] != 2022].copy()

# 回答者数を数値化
df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)

# =========================
# 2. 区分ごとのデータ数を集計
# =========================
summary = (
    df.groupby("区分")
    .agg(
        総行数=("回答者数", "size"),
        回答者数0の件数=(
            "回答者数",
            lambda x: (x == 0).sum()
        ),
        使用できるデータ数=(
            "回答者数",
            lambda x: (x > 0).sum()
        )
    )
    .reset_index()
)

# 割合を計算
summary["回答者数0の割合(%)"] = (
    summary["回答者数0の件数"]
    / summary["総行数"]
    * 100
).round(1)

summary["使用できるデータの割合(%)"] = (
    summary["使用できるデータ数"]
    / summary["総行数"]
    * 100
).round(1)

# 列の順番を整理
summary = summary[
    [
        "区分",
        "総行数",
        "回答者数0の件数",
        "回答者数0の割合(%)",
        "使用できるデータ数",
        "使用できるデータの割合(%)"
    ]
]

display(summary)
```

## 年代、性別でどれだけが差があるかを調べる

```python
import pandas as pd
import numpy as np

# =========================
# 1. データ準備
# =========================
df = df_sheet2.copy()

# 2022年度を除外
df = df[df["年度"] != 2022].copy()

# 数値化
df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)

df["幸福度"] = pd.to_numeric(
    df["幸福度"],
    errors="coerce"
)

# 24カテゴリー
subjective_cols = [
    "主観_医療・福祉",
    "主観_買物・飲食",
    "主観_住宅環境",
    "主観_移動・交通",
    "主観_遊び・娯楽",
    "主観_子育て",
    "主観_初等・中等教育",
    "主観_地域行政",
    "主観_デジタル生活",
    "主観_公共空間",
    "主観_都市景観",
    "主観_事故・犯罪",
    "主観_自然景観",
    "主観_自然の恵み",
    "主観_環境共生",
    "主観_自然災害",
    "主観_地域とのつながり",
    "主観_多様性と寛容性",
    "主観_自己効力感",
    "主観_健康状態",
    "主観_文化・芸術",
    "主観_教育機会の豊かさ",
    "主観_雇用・所得",
    "主観_事業創造"
]

target_cols = ["幸福度"] + subjective_cols


# =========================
# 2. 回答者数で重み付けした平均を計算する関数
# =========================
def weighted_mean(group, value_col):
    temp = group[[value_col, "回答者数"]].dropna()

    # 回答者数が0以下のものは除外
    temp = temp[temp["回答者数"] > 0]

    if len(temp) == 0:
        return np.nan

    return np.average(
        temp[value_col],
        weights=temp["回答者数"]
    )


# =========================
# 3. 年代別の比較
# =========================
df_age = df[
    (df["区分"] == "年代×性別")
].copy()

# 性別に関係なく年代ごとの差を見る
age_table = pd.DataFrame({
    col: df_age.groupby("年代").apply(
        lambda g, c=col: weighted_mean(g, c),
        include_groups=False
    )
    for col in target_cols
})

print("===== 年代別の平均 =====")
display(age_table.round(2))


# =========================
# 4. 性別別の比較
# =========================
gender_table = pd.DataFrame({
    col: df_age.groupby("性別").apply(
        lambda g, c=col: weighted_mean(g, c),
        include_groups=False
    )
    for col in target_cols
})

print("===== 性別別の平均 =====")
display(gender_table.round(2))


# =========================
# 5. 年代×性別の比較
# =========================
age_gender_table = pd.DataFrame({
    col: df_age.groupby(["年代", "性別"]).apply(
        lambda g, c=col: weighted_mean(g, c),
        include_groups=False
    )
    for col in target_cols
})

print("===== 年代×性別の平均 =====")
display(age_gender_table.round(2))
```

```python
# =========================
# 6. 年代による差の大きさ
# =========================
age_difference = (
    age_table.max() - age_table.min()
).sort_values(ascending=False)

print("===== 年代による最大差 =====")
display(
    age_difference
    .rename("最大値－最小値")
    .round(2)
)


# =========================
# 7. 性別による差の大きさ
# =========================
gender_difference = (
    gender_table.max() - gender_table.min()
).sort_values(ascending=False)

print("===== 性別による最大差 =====")
display(
    gender_difference
    .rename("最大値－最小値")
    .round(2)
)


# =========================
# 8. 年代×性別による差の大きさ
# =========================
age_gender_difference = (
    age_gender_table.max() - age_gender_table.min()
).sort_values(ascending=False)

print("===== 年代×性別による最大差 =====")
display(
    age_gender_difference
    .rename("最大値－最小値")
    .round(2)
)
```

```python
import pandas as pd

# =========================
# 1. データ準備
# =========================
df = df_sheet2.copy()

# 2022年度を除外
df = df[df["年度"] != 2022].copy()

# 数値化
df["幸福度"] = pd.to_numeric(
    df["幸福度"],
    errors="coerce"
)

df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)


# =========================
# 2. 区分ごとの幸福度欠損数・有効データ数
# =========================
summary = (
    df.groupby("区分")
    .agg(
        総データ数=("幸福度", "size"),
        幸福度の欠損数=("幸福度", lambda x: x.isna().sum()),
        幸福度あり=("幸福度", lambda x: x.notna().sum()),
        幸福度なし_回答者数1以上=(
            "幸福度",
            lambda x: (
                x.isna()
            ).sum()
        )
    )
    .reset_index()
)

# 「幸福度なし かつ 回答者数1以上」を別途正確に集計
missing_with_respondents = (
    df[
        (df["幸福度"].isna()) &
        (df["回答者数"] >= 1)
    ]
    .groupby("区分")
    .size()
    .rename("幸福度なし_回答者数1以上")
    .reset_index()
)

summary = summary.drop(
    columns=["幸福度なし_回答者数1以上"]
).merge(
    missing_with_respondents,
    on="区分",
    how="left"
)

summary["幸福度なし_回答者数1以上"] = (
    summary["幸福度なし_回答者数1以上"]
    .fillna(0)
    .astype(int)
)

summary["幸福度の欠損率(%)"] = (
    summary["幸福度の欠損数"]
    / summary["総データ数"]
    * 100
).round(1)

display(summary)


# =========================
# 3. 区分ごとの詳細確認
# =========================
for kubun in df["区分"].dropna().unique():

    print(f"\n{'=' * 70}")
    print(f"区分：{kubun}")
    print(f"{'=' * 70}")

    # -------------------------
    # 幸福度あり
    # -------------------------
    valid = df[
        (df["区分"] == kubun) &
        (df["幸福度"].notna())
    ].copy()

    valid = valid.sort_values(
        ["年度", "都道府県", "市区町村"]
    )

    examples_valid = (
        valid[
            [
                "年度",
                "都道府県",
                "市区町村",
                "地域ID",
                "年代",
                "性別",
                "回答者数",
                "幸福度"
            ]
        ]
        .drop_duplicates(
            subset=["都道府県", "市区町村"]
        )
        .head(5)
    )

    print(f"\n【幸福度あり】{len(valid):,}件")
    display(examples_valid)

    # -------------------------
    # 幸福度なし（全て）
    # -------------------------
    missing = df[
        (df["区分"] == kubun) &
        (df["幸福度"].isna())
    ].copy()

    missing = missing.sort_values(
        ["年度", "都道府県", "市区町村"]
    )

    examples_missing = (
        missing[
            [
                "年度",
                "都道府県",
                "市区町村",
                "地域ID",
                "年代",
                "性別",
                "回答者数"
            ]
        ]
        .drop_duplicates(
            subset=["都道府県", "市区町村"]
        )
        .head(5)
    )

    print(f"\n【幸福度なし】{len(missing):,}件")
    display(examples_missing)

    # -------------------------
    # 幸福度なし ＋ 回答者数1以上
    # -------------------------
    missing_with_n = missing[
        missing["回答者数"] >= 1
    ].copy()

    examples_missing_with_n = (
        missing_with_n[
            [
                "年度",
                "都道府県",
                "市区町村",
                "地域ID",
                "年代",
                "性別",
                "回答者数"
            ]
        ]
        .drop_duplicates(
            subset=["都道府県", "市区町村"]
        )
        .head(5)
    )

    print(
        f"\n【幸福度なし ＋ 回答者数1以上】"
        f"{len(missing_with_n):,}件"
    )
    display(examples_missing_with_n)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from pathlib import Path

# =========================
# 1. 日本語フォントを設定
# =========================
jp_fonts = [
    "Noto Sans CJK JP",
    "Noto Sans JP",
    "IPAexGothic",
    "IPAGothic",
    "Yu Gothic",
    "Meiryo"
]

available_fonts = {f.name for f in fm.fontManager.ttflist}

for font_name in jp_fonts:
    if font_name in available_fonts:
        plt.rcParams["font.family"] = font_name
        break

# マイナス記号の文字化け対策
plt.rcParams["axes.unicode_minus"] = False


# =========================
# 2. 分析対象を作成
# =========================
df = df_sheet2.copy()

df["回答者数"] = pd.to_numeric(
    df["回答者数"],
    errors="coerce"
)

df["幸福度"] = pd.to_numeric(
    df["幸福度"],
    errors="coerce"
)

# 2023～2026年
# 区分 = 全体
# 幸福度あり
# 回答者数 > 0
df_analysis = df[
    (df["年度"].between(2023, 2026)) &
    (df["区分"] == "全体") &
    (df["幸福度"].notna()) &
    (df["回答者数"] > 0)
].copy()


# =========================
# 3. 回答者数を取得
# =========================
n = df_analysis["回答者数"].dropna()


# =========================
# 4. グラフ作成
# =========================
fig, ax = plt.subplots(figsize=(10, 6))

# 回答者数の分布
bins = np.logspace(
    np.log10(n.min()),
    np.log10(n.max()),
    30
)

ax.hist(
    n,
    bins=bins
)

# 基準候補を表示
thresholds = [10, 30, 50, 80, 100]
threshold_colors = ["red", "orange", "green", "blue", "purple"]

for threshold, color in zip(thresholds, threshold_colors):
    ax.axvline(
        threshold,
        linestyle="--",
        color=color,
        label=f"n = {threshold}"
    )

# 対数軸
ax.set_xscale("log")

ax.set_xlabel("回答者数")
ax.set_ylabel("観測数")

ax.set_title(
    f"回答者数の分布（2023～2026年・全体区分、n = {len(n):,}）"
)

ax.legend()
ax.grid(
    True,
    which="both",
    axis="both",
    alpha=0.3
)

plt.tight_layout()
# =========================
# 5. nの基準値ごとの件数・割合
# =========================

thresholds = [10, 30, 50, 80, 100]

results = []

total = len(df_analysis)

for threshold in thresholds:
    count_ge = (df_analysis["回答者数"] >= threshold).sum()
    count_lt = (df_analysis["回答者数"] < threshold).sum()

    results.append({
        "基準値": threshold,
        "n以上の観測数": count_ge,
        "n以上の割合(%)": count_ge / total * 100,
        "n未満の観測数": count_lt,
        "n未満の割合(%)": count_lt / total * 100
    })

threshold_summary = pd.DataFrame(results)

threshold_summary["n以上の割合(%)"] = threshold_summary["n以上の割合(%)"].round(1)
threshold_summary["n未満の割合(%)"] = threshold_summary["n未満の割合(%)"].round(1)

display(threshold_summary)
```
